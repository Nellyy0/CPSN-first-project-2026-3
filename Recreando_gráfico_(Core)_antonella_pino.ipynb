```python
import seaborn as sns
import matplotlib.pyplot as plt


sns.set_theme(style="darkgrid")
sns.set_palette(["#E24A33", "#348ABD"])

df = sns.load_dataset("tips")


fig, ax = plt.subplots(figsize=(12, 6))


sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    hue_order=["Lunch", "Dinner"],
    style="smoker",
    style_order=["Yes", "No"],
    markers={"Yes": "o", "No": "X"},
    s=100,
    ax=ax
)

ax.set_title(
    "Relación entre Total Bill y Tip con Anotaciones y Flechas",
    fontsize=20
)
ax.set_xlabel("Total Bill ($)", fontsize=15)
ax.set_ylabel("Tip ($)", fontsize=15)

ax.legend(
    loc="upper right",
    fontsize="large",
    shadow=True,
    title="Detalles"
)


ax.grid(
    True,
    which="major",
    linestyle="--",
    linewidth=0.5,
    color="gray"
)


ax.annotate(
    "Mayor tip en la cena",
    xy=(50, 10),
    xytext=(40, 15),
    arrowprops=dict(
        facecolor="black",
        shrink=0.05
    )
)

ax.arrow(
    30, 5,
    10, 3,
    head_width=0.8,
    head_length=0.5,
    fc="green",
    ec="green"
)

plt.show()
```
