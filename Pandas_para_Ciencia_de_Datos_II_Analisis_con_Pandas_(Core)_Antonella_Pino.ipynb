```python
import pandas as pd
from google.colab import files
from IPython.display import display


print("Selecciona el archivo vgsales.csv.")
archivos = files.upload()

if "vgsales.csv" not in archivos:
    raise ValueError("Debes subir el archivo llamado vgsales.csv.")

datos = pd.read_csv("vgsales.csv")

print("\nPRIMERAS 10 FILAS")
display(datos.head(10))


print("\nÚLTIMAS 5 FILAS")
display(datos.tail(5))

print("\nINFORMACIÓN GENERAL")
datos.info()


print("\nESTADÍSTICAS DESCRIPTIVAS")
display(datos.describe())


print("\nTIPOS DE DATOS DE CADA COLUMNA")
print(datos.dtypes)


print("\nCANTIDAD DE REGISTROS POR GÉNERO")
print(datos["Genre"].value_counts())


print("\nPLATAFORMAS DISPONIBLES")
print(datos["Platform"].unique())


ventas_norteamerica = datos[datos["NA_Sales"] > 1]

print("\nVENTAS EN AMÉRICA DEL NORTE MAYORES A 1 MILLÓN")
print("Cantidad de registros:", len(ventas_norteamerica))
display(ventas_norteamerica)


ventas_japon = datos[datos["JP_Sales"] < 0.1]

print("\nVENTAS EN JAPÓN MENORES A 0.1 MILLÓN")
print("Cantidad de registros:", len(ventas_japon))
display(ventas_japon)


juegos_accion = datos.query(
    "Genre == 'Action' and Global_Sales > 2"
)

print("\nJUEGOS DE ACCIÓN CON VENTAS GLOBALES MAYORES A 2 MILLONES")
print("Cantidad de registros:", len(juegos_accion))
display(juegos_accion)


print("\nCOLUMNAS NAME Y GLOBAL_SALES")
display(datos[["Name", "Global_Sales"]])

print("\nFILAS CON ÍNDICES DEL 5 AL 10 Y COLUMNAS NAME Y GENRE")
display(datos.loc[5:10, ["Name", "Genre"]])


print("\nPRIMERAS 5 FILAS Y PRIMERAS 3 COLUMNAS")
display(datos.iloc[0:5, 0:3])
```
